## Ejercicio 1
Conversión de segundos a días, horas, minutos y segundos.


```python
# Ejercicio 1

def is_integer(text: str) -> bool:
    """Comprueba si una cadena representa un número entero."""
    try:
        int(text)
        return True
    except ValueError:
        return False


def seconds_to_dhms(total_seconds: int) -> tuple[int, int, int, int]:
    """Convierte una cantidad de segundos en días, horas, minutos y segundos."""

    days = total_seconds // 86400
    remaining_seconds = total_seconds % 86400

    hours = remaining_seconds // 3600
    remaining_seconds = remaining_seconds % 3600

    minutes = remaining_seconds // 60
    seconds = remaining_seconds % 60

    return days, hours, minutes, seconds


def main() -> None:
    """Solicita una cantidad de segundos y muestra su conversión."""

    while True:
        text = input("Introduzca los segundos: ")

        if is_integer(text):
            total_seconds = int(text)

            if total_seconds >= 0:
                break

        print("Entrada no válida. Introduzca un entero no negativo.")

    days, hours, minutes, seconds = seconds_to_dhms(total_seconds)

    print(
        f"{total_seconds} segundos son "
        f"{days} d, {hours} h, {minutes} min y {seconds} s."
    )


main()
```

## Ejercicio 2
Comprobación y búsqueda de años bisiestos.

```python
# Ejercicio 2

def is_leap_year(year: int) -> bool:
    """Devuelve True si el año es bisiesto y False en caso contrario."""
    return (year % 4 == 0 and year % 100 != 0) or year % 400 == 0

def leap_years_between(start: int, end: int) -> list[int]:
    """Devuelve una lista con los años bisiestos entre start y end, ambos incluidos."""
    leap_years = []

    for year in range(start, end + 1):
        if is_leap_year(year):
            leap_years.append(year)

    return leap_years

print(leap_years_between(1896, 1912))

def explain_leap_year(year: int) -> str:
    """Devuelve una explicación de por qué un año es o no bisiesto."""

    if year % 400 == 0:
        return f"{year} es bisiesto: es divisible entre 400."

    if year % 100 == 0:
        return f"{year} no es bisiesto: es divisible entre 100 pero no entre 400."

    if year % 4 == 0:
        return f"{year} es bisiesto: es divisible entre 4 y no entre 100."

    return f"{year} no es bisiesto: no es divisible entre 4."

def main() -> None:
    """Solicita un año y muestra si es bisiesto y el motivo."""

    year = int(input("Año: "))

    if year < 1582:
        print("El calendario gregoriano se estableció en 1582.")
    else:
        print(explain_leap_year(year))
main()

```

## Ejercicio 3
Procesamiento de cadenas y sustitución de palabras.

```python
# Ejercicio 3

def process_text(
    text: str = "This is a default text with Python and amazing words.",
    words: list[str] = ["Python", "amazing"]
) -> tuple[str, int]:
    """Procesa un texto y reemplaza las palabras indicadas por asteriscos."""

    text = text.lower().strip()
    words = [word.lower() for word in words]

    word_list = text.split()
    count = 0

    for i in range(len(word_list)):
        if word_list[i] in words:
            word_list[i] = "*" * len(word_list[i])
            count += 1

    processed_text = " ".join(word_list)

    return processed_text, count


def main() -> None:
    """Prueba la función process_text con los valores predeterminados."""

    processed_text, count = process_text()

    print(f'Processed text: "{processed_text}"')
    print(f"Palabras reemplazadas: {count}")


main()
```

## Ejercicio 4 - Cifrado César

El cifrado César consiste en desplazar cada letra de un texto un número determinado de posiciones (`shift`) dentro del alfabeto.

```python
# Ejercicio 4

def shift_char(char: str, shift: int) -> str:
    """Desplaza una letra según el cifrado César."""

    shift = shift % 26

    if 'a' <= char <= 'z':
        return chr((ord(char) - ord('a') + shift) % 26 + ord('a'))

    if 'A' <= char <= 'Z':
        return chr((ord(char) - ord('A') + shift) % 26 + ord('A'))

    return char


def encrypt(text: str, shift: int) -> str:
    """Cifra un texto utilizando el cifrado César."""

    encrypted_text = ""

    for char in text:
        encrypted_text += shift_char(char, shift)

    return encrypted_text


def decrypt(text: str, shift: int) -> str:
    """Descifra un texto cifrado mediante el cifrado César."""
    return encrypt(text, -shift)


def brute_force(cipher_text: str) -> list[str]:
    """Devuelve los 26 posibles descifrados de un texto."""

    results = []

    for shift in range(26):
        results.append(decrypt(cipher_text, shift))

    return results


def main() -> None:
    """Prueba las funciones del cifrado César."""

    print(encrypt("Hola, Mundo!", 3))
    print(decrypt("Krod, Pxqgr!", 3))
    print(encrypt("xyz", 3))
    print(encrypt("abc", -3))
    print(brute_force("Krod, Pxqgr!")[3])

    # Caso adicional
    print(encrypt("Python 123!", 29))


main()
```

## Ejercicio 5 - Palíndromos

En este ejercicio se implementan distintas funciones para detectar palíndromos, ignorando espacios, signos de puntuación, mayúsculas y tildes.

```python
# Ejercicio 5

def normalize(text: str) -> str:
    """Normaliza un texto para poder comprobar si es un palíndromo."""

    text = text.lower()

    text = text.replace("á", "a")
    text = text.replace("é", "e")
    text = text.replace("í", "i")
    text = text.replace("ó", "o")
    text = text.replace("ú", "u")
    text = text.replace("ü", "u")

    normalized_text = ""

    for char in text:
        if char.isalnum():
            normalized_text += char

    return normalized_text


def is_palindrome(text: str) -> bool:
    """Devuelve True si el texto es un palíndromo."""

    normalized_text = normalize(text)

    return normalized_text == normalized_text[::-1]


def find_palindromes(sentences: list[str]) -> list[str]:
    """Devuelve las frases de una lista que son palíndromos."""

    palindromes = []

    for sentence in sentences:
        if is_palindrome(sentence):
            palindromes.append(sentence)

    return palindromes


def palindrome_words(text: str) -> list[str]:
    """Devuelve las palabras palíndromas de al menos 3 letras."""

    palindromes = []

    for word in text.split():
        if len(word) >= 3 and is_palindrome(word):
            palindromes.append(word)

    return palindromes


def main() -> None:
    """Prueba las funciones relacionadas con palíndromos."""

    sentences = [
        "Anita lava la tina",
        "Dábale arroz a la zorra el abad",
        "Esto no es un palíndromo",
        "¿Acaso hubo búhos acá?",
    ]

    for phrase in find_palindromes(sentences):
        print(phrase)

    print(palindrome_words("Ana vio un oso en el ojo de Anita"))

    # Caso adicional
    print(is_palindrome("Reconocer"))


main()
```

## Ejercicio 6 - Análisis de números primos

En este ejercicio se implementan distintas funciones para trabajar con números primos.

```python
# Ejercicio 6

def is_prime(n: int) -> bool:
    """Comprueba si un número es primo."""

    if n < 2:
        return False

    for divisor in range(2, n):
        if n % divisor == 0:
            return False

    return True


def prime_list(limit: int) -> list[int]:
    """Genera una lista de números primos hasta el límite incluido."""

    primes = []

    for number in range(2, limit + 1):
        if is_prime(number):
            primes.append(number)

    return primes


def check_palindrome(primes: list[int]) -> list[int]:
    """Devuelve los números primos que son palíndromos."""

    palindrome_primes = []

    for prime in primes:
        text = str(prime)

        if text == text[::-1]:
            palindrome_primes.append(prime)

    return palindrome_primes


def categorize_prime(prime: int) -> str:
    """Clasifica un número primo según su tamaño."""

    if prime < 10:
        return "pequeño"

    if prime < 100:
        return "mediano"

    return "grande"


def main() -> None:
    """Prueba las funciones de análisis de números primos."""

    limit = 200

    primes = prime_list(limit)
    print("Números primos:", primes)

    palindrome_primes = check_palindrome(primes)
    print("Primos palíndromos:", palindrome_primes)

    print("Categoría de 7:", categorize_prime(7))
    print("Categoría de 47:", categorize_prime(47))
    print("Categoría de 101:", categorize_prime(101))


main()
```

## Ejercicio 7 - Operaciones con listas

En este ejercicio se implementan diferentes operaciones sobre listas de números enteros sin utilizar `sorted()`, `list.sort()`, `set()` ni el módulo `statistics`.

```python
# Ejercicio 7

def remove_duplicates(numbers: list[int]) -> list[int]:
    """Devuelve una nueva lista sin elementos repetidos."""

    result = []

    for number in numbers:
        if number not in result:
            result.append(number)

    return result


def rotate_left(numbers: list[int], k: int) -> list[int]:
    """Devuelve una nueva lista rotada k posiciones hacia la izquierda."""

    if len(numbers) == 0:
        return []

    k = k % len(numbers)

    return numbers[k:] + numbers[:k]


def split_even_odd(numbers: list[int]) -> tuple[list[int], list[int]]:
    """Separa los números pares e impares manteniendo su orden."""

    evens = []
    odds = []

    for number in numbers:
        if number % 2 == 0:
            evens.append(number)
        else:
            odds.append(number)

    return evens, odds


def merge_sorted(a: list[int], b: list[int]) -> list[int]:
    """Combina dos listas ordenadas en una única lista ordenada."""

    result = []
    i = 0
    j = 0

    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1

    while i < len(a):
        result.append(a[i])
        i += 1

    while j < len(b):
        result.append(b[j])
        j += 1

    return result


def second_largest(numbers: list[int]) -> int | None:
    """Devuelve el segundo valor más grande distinto del máximo."""

    largest = None
    second = None

    for number in numbers:
        if largest is None or number > largest:
            second = largest
            largest = number

        elif number != largest and (second is None or number > second):
            second = number

    return second


def main() -> None:
    """Prueba las funciones del ejercicio."""

    print(remove_duplicates([3, 1, 3, 2, 1]))

    print(rotate_left([1, 2, 3, 4, 5], 2))
    print(rotate_left([1, 2, 3, 4, 5], 7))
    print(rotate_left([1, 2, 3, 4, 5], -1))
    print(rotate_left([1, 2, 3, 4, 5], 0))

    print(split_even_odd([1, 2, 3, 4, 5, 6]))

    print(merge_sorted([1, 4, 9], [2, 3, 10, 11]))

    print(second_largest([4, 9, 2, 9, 7]))
    print(second_largest([5, 5]))

    # Caso adicional
    print(second_largest([10, 20, 30, 40]))


main()
```

## Ejercicio 8 - Menú de operaciones con listas

En este ejercicio se implementa un programa que realiza diferentes operaciones sobre una lista de números enteros mediante un menú.

```python
# Ejercicio 8

def sum_numbers(numbers: list[int]) -> int:
    """Devuelve la suma de todos los números de la lista."""

    total = 0

    for number in numbers:
        total += number

    return total


def sum_positive_negative(numbers: list[int]) -> tuple[int, int]:
    """Devuelve por separado la suma de positivos y negativos."""

    positive_sum = 0
    negative_sum = 0

    for number in numbers:
        if number > 0:
            positive_sum += number
        elif number < 0:
            negative_sum += number

    return positive_sum, negative_sum


def max_min(numbers: list[int]) -> tuple[int, int] | None:
    """Devuelve el máximo y el mínimo o None si la lista está vacía."""

    if len(numbers) == 0:
        return None

    maximum = numbers[0]
    minimum = numbers[0]

    for number in numbers:
        if number > maximum:
            maximum = number

        if number < minimum:
            minimum = number

    return maximum, minimum


def all_unique(numbers: list[int]) -> bool:
    """Devuelve True si todos los números son únicos."""

    seen = []

    for number in numbers:
        if number in seen:
            return False

        seen.append(number)

    return True


def compare_positive_negative(numbers: list[int]) -> str:
    """Indica si hay más números positivos, negativos o la misma cantidad."""

    positives = 0
    negatives = 0

    for number in numbers:
        if number > 0:
            positives += 1
        elif number < 0:
            negatives += 1

    return (
        "Hay más números positivos."
        if positives > negatives
        else "Hay más números negativos."
        if negatives > positives
        else "Hay tantos números positivos como negativos."
    )


def main() -> None:
    """Ejecuta el menú de operaciones sobre una lista de números."""

    text = input("Introduzca números enteros separados por espacios: ")

    if text.strip() == "":
        numbers = []
    else:
        numbers = [int(number) for number in text.split()]

    while True:
        print("\nMENÚ")
        print("1. Suma de todos los números")
        print("2. Suma de positivos y negativos")
        print("3. Máximo y mínimo")
        print("4. Comprobar si todos son únicos")
        print("5. Comparar cantidad de positivos y negativos")
        print("6. Salir")

        option = input("Seleccione una opción: ")

        if option == "1":
            print("Suma total:", sum_numbers(numbers))

        elif option == "2":
            positive_sum, negative_sum = sum_positive_negative(numbers)
            print("Suma de positivos:", positive_sum)
            print("Suma de negativos:", negative_sum)

        elif option == "3":
            result = max_min(numbers)

            if result is None:
                print("La lista está vacía. No existe máximo ni mínimo.")
            else:
                maximum, minimum = result
                print("Máximo:", maximum)
                print("Mínimo:", minimum)

        elif option == "4":
            print("Todos los números son únicos:", all_unique(numbers))

        elif option == "5":
            print(compare_positive_negative(numbers))

        elif option == "6":
            print("Fin del programa.")
            break

        else:
            print("Opción no válida.")


main()
```

## Ejercicio 9 - Polinomios

En este ejercicio se representan polinomios mediante listas de coeficientes, donde la posición de cada coeficiente indica el exponente correspondiente.

```python
# Ejercicio 9

def format_monomial(coef: int | float, exp: int) -> str:
    """Devuelve un monomio como cadena con su signo explícito."""

    if coef == 0:
        return ""

    sign = "+" if coef > 0 else "-"
    abs_coef = abs(coef)

    if exp == 0:
        return f"{sign}{abs_coef}"

    if abs_coef == 1:
        coefficient = ""
    else:
        coefficient = str(abs_coef)

    if exp == 1:
        return f"{sign}{coefficient}x"

    return f"{sign}{coefficient}x**{exp}"


def format_polynomial(coefs: list[int | float]) -> str:
    """Devuelve el polinomio completo como una cadena."""

    terms = []

    for exp in range(len(coefs) - 1, -1, -1):
        monomial = format_monomial(coefs[exp], exp)

        if monomial != "":
            terms.append(monomial)

    if len(terms) == 0:
        return "0"

    polynomial = terms[0]

    if polynomial.startswith("+"):
        polynomial = polynomial[1:]

    for term in terms[1:]:
        if term.startswith("+"):
            polynomial += " + " + term[1:]
        else:
            polynomial += " - " + term[1:]

    return polynomial


def evaluate(coefs: list[int | float], x: int | float) -> int | float:
    """Calcula el valor del polinomio para un valor de x."""

    result = 0

    for exp in range(len(coefs)):
        result += coefs[exp] * x ** exp

    return result


def derivative(coefs: list[int | float]) -> list[int | float]:
    """Devuelve los coeficientes del polinomio derivado."""

    derivative_coefs = []

    for exp in range(1, len(coefs)):
        derivative_coefs.append(coefs[exp] * exp)

    return derivative_coefs


def main() -> None:
    """Prueba las funciones relacionadas con polinomios."""

    print(format_monomial(2.0, 3))
    print(format_monomial(1.0, 2))
    print(format_monomial(-1.0, 1))
    print(format_monomial(4.0, 0))
    print(format_monomial(0.0, 5))

    print(format_polynomial([4.0, 1.0, 0.0, 2.0]))
    print(format_polynomial([0.0, -1.0, 3.0]))
    print(format_polynomial([-2.5, 0.0, -1.0]))
    print(format_polynomial([0.0, 0.0]))

    print(evaluate([4.0, 1.0, 0.0, 2.0], 2))
    print(derivative([4.0, 1.0, 0.0, 2.0]))

    # Caso adicional
    print(format_polynomial([5, -2, 1]))
    print(derivative([5, -2, 1]))


main()
```

## Ejercicio 10 - Operaciones con cadenas

En este ejercicio se implementa un programa que realiza diferentes operaciones sobre una lista de cadenas mediante un menú.

```python
# Ejercicio 10

def strings_starting_with(strings: list[str], letter: str) -> list[str]:
    """Devuelve las cadenas que comienzan por la letra indicada."""

    result = []

    for text in strings:
        if text.startswith(letter):
            result.append(text)

    return result


def count_substring(strings: list[str], substring: str) -> int:
    """Cuenta cuántas cadenas contienen una subcadena."""

    count = 0

    for text in strings:
        if substring in text:
            count += 1

    return count


def longest_shortest(strings: list[str]) -> tuple[str, str] | None:
    """Devuelve la cadena más larga y la más corta."""

    if len(strings) == 0:
        return None

    longest = strings[0]
    shortest = strings[0]

    for text in strings[1:]:
        if len(text) > len(longest):
            longest = text

        if len(text) < len(shortest):
            shortest = text

    return longest, shortest


def same_object(first: str, second: str) -> bool:
    """Comprueba si dos cadenas son el mismo objeto."""
    return first is second


def check_long_string(strings: list[str]) -> None:
    """Comprueba con for...else si alguna cadena tiene más de 10 caracteres."""

    for text in strings:
        if len(text) > 10:
            print("Hay al menos una cadena con más de 10 caracteres.")
            break
    else:
        print("Ninguna cadena tiene más de 10 caracteres.")


def main() -> None:
    """Ejecuta el menú de operaciones sobre cadenas."""

    text = input("Introduzca cadenas separadas por comas: ")

    if text.strip() == "":
        strings = []
    else:
        strings = [item.strip() for item in text.split(",")]

    while True:
        print("\nMENÚ")
        print("1. Cadenas que comienzan por una letra")
        print("2. Contar cadenas que contienen una subcadena")
        print("3. Cadena más larga y más corta")
        print("4. Comprobar si dos cadenas son el mismo objeto")
        print("5. Comprobar cadenas con más de 10 caracteres")
        print("6. Salir")

        option = input("Seleccione una opción: ")

        if option == "1":
            letter = input("Introduzca una letra: ")

            for item in strings_starting_with(strings, letter):
                print(item)

        elif option == "2":
            substring = input("Introduzca una subcadena: ")
            print("Cantidad:", count_substring(strings, substring))

        elif option == "3":
            result = longest_shortest(strings)

            if result is None:
                print("La lista está vacía.")
            else:
                longest, shortest = result
                print("Cadena más larga:", longest)
                print("Cadena más corta:", shortest)

        elif option == "4":
            if len(strings) < 2:
                print("Se necesitan al menos dos cadenas.")
            else:
                print(
                    "¿Son el mismo objeto?",
                    same_object(strings[0], strings[1])
                )

        elif option == "5":
            check_long_string(strings)

        elif option == "6":
            print("Fin del programa.")
            break

        else:
            print("Opción no válida.")


main()
```

```python
## Ejercicio 11 - Anonimización de informes médicos

En este ejercicio se utilizan expresiones regulares mediante el módulo `re` para anonimizar los datos sensibles de un informe médico.
```

```python
# Ejercicio 11

import re


date_pattern = re.compile(r"\d{1,2}/\d{1,2}/\d{4}")


def calculate_age(birth_date: str, consultation_date: str) -> int:
    """Calcula la edad del paciente en la fecha de consulta."""

    birth_day, birth_month, birth_year = map(int, birth_date.split("/"))
    consultation_day, consultation_month, consultation_year = map(
        int, consultation_date.split("/")
    )

    age = consultation_year - birth_year

    if (consultation_month, consultation_day) < (birth_month, birth_day):
        age -= 1

    return age


def anonymize_report(report: str) -> str:
    """Anonimiza los datos sensibles de un informe médico."""

    birth_match = re.search(
        r"^Fecha de nacimiento:\s*(\d{1,2}/\d{1,2}/\d{4})",
        report,
        re.MULTILINE
    )

    consultation_match = re.search(
        r"^(\d{1,2}/\d{1,2}/\d{4})\s+(?:Dr\.|Dra\.)",
        report,
        re.MULTILINE
    )

    if birth_match and consultation_match:
        birth_date = birth_match.group(1)
        consultation_date = consultation_match.group(1)

        age = calculate_age(birth_date, consultation_date)

        report = re.sub(
            r"^Fecha de nacimiento:\s*\d{1,2}/\d{1,2}/\d{4}",
            f"Edad: {age}",
            report,
            flags=re.MULTILINE
        )

    report = re.sub(
        r"^Nombre:.*$",
        "Nombre: PACIENTE",
        report,
        flags=re.MULTILINE
    )

    report = date_pattern.sub("FECHA", report)

    report = re.sub(
        r"(?:Dr\.|Dra\.).*$",
        "MEDICO",
        report,
        flags=re.MULTILINE
    )

    return report


def main() -> None:
    """Prueba la anonimización de un informe médico."""

    informe = """Informe clínico de Urgencias
Nombre: Juan Pérez López
Género: Hombre
Fecha de nacimiento: 12/05/1980
Motivo de consulta: dolor abdominal
12/05/2023 Dr. Ramírez
Tratamiento: reposo y analgésicos"""

    print(anonymize_report(informe))


main()
```

## Ejercicio 12 - Anagramas

En este ejercicio se implementan diferentes funciones para trabajar con palabras que son anagramas entre sí.

```python
# Ejercicio 12

def is_anagram(word1: str, word2: str) -> bool:
    """Comprueba si dos cadenas son anagramas ignorando mayúsculas y espacios."""

    word1 = word1.lower().replace(" ", "")
    word2 = word2.lower().replace(" ", "")

    return sorted(word1) == sorted(word2)


def group_anagrams(words: list[str]) -> list[list[str]]:
    """Agrupa las palabras que son anagramas manteniendo su orden."""

    groups = []

    for word in words:
        added = False

        for group in groups:
            if is_anagram(word, group[0]):
                group.append(word)
                added = True
                break

        if not added:
            groups.append([word])

    return groups


def largest_group(groups: list[list[str]]) -> list[str]:
    """Devuelve el grupo con más palabras."""

    if len(groups) == 0:
        return []

    largest = groups[0]

    for group in groups[1:]:
        if len(group) > len(largest):
            largest = group

    return largest


def find_anagrams(word: str, words: list[str]) -> list[str]:
    """Devuelve los anagramas de una palabra excluyendo las palabras iguales."""

    result = []

    for candidate in words:
        if candidate.lower() != word.lower() and is_anagram(word, candidate):
            result.append(candidate)

    return result


def main() -> None:
    """Prueba las funciones relacionadas con anagramas."""

    words = ["roma", "amor", "pelo", "mora", "casa", "lope", "ramo"]

    print(is_anagram("Roma", "amor"))
    print(is_anagram("casa", "caso"))
    print(is_anagram("dormitory", "dirty room"))

    groups = group_anagrams(words)
    print(groups)

    print(largest_group(groups))

    print(find_anagrams("amor", words))
    print(find_anagrams("amor", ["amor", "roma", "AMOR", "roma"]))
    print(find_anagrams("sol", words))

    # Caso adicional
    print(is_anagram("listen", "silent"))


main()
```
